```python
import math

diametro = 10
radio = diametro / 2
area = math.pi * radio ** 2
print(f"El área del círculo es: {area:.2f}")
```

```python
import math

def area_circulo(radio):
    """Calcula el área de un círculo dado su radio."""
    return math.pi * radio ** 2

# Uso
print(area_circulo(5))  # 78.53981633974483
```

```python
import numpy as np

radio = 5
area = np.pi * radio ** 2
print(f"El área del círculo es: {area:.2f}")
```

```python
import matplotlib.pyplot as plt
import numpy as np

radio = 5
area = np.pi * radio ** 2

fig, ax = plt.subplots()
circulo = plt.Circle((0, 0), radio, fill=False, color='blue', linewidth=2)
ax.add_patch(circulo)
ax.plot([0, radio], [0, 0], 'r--')  # línea del radio
ax.set_xlim(-radio-1, radio+1)
ax.set_ylim(-radio-1, radio+1)
ax.set_aspect('equal')
ax.set_title(f"Círculo con radio {radio}\nÁrea = {area:.2f}")
plt.show()
```
