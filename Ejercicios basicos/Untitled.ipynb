```python
import matplotlib.pyplot as plt

lado = 5
area = lado ** 2

# Crear la figura
fig, ax = plt.subplots()
cuadrado = plt.Rectangle((0, 0), lado, lado, fill=True, color='skyblue', edgecolor='black')
ax.add_patch(cuadrado)

# Ajustar los límites y aspecto
ax.set_xlim(-1, lado + 1)
ax.set_ylim(-1, lado + 1)
ax.set_aspect('equal')
ax.set_title(f'Cuadrado de lado {lado} — Área: {area}')

plt.grid(True)
plt.show()
```

```python
import math
import matplotlib.pyplot as plt

# Datos
radio = 5

# Cálculo del área
area = math.pi * radio**2

# Mostrar resultado
print(f"El radio del círculo es: {radio}")
print(f"El área del círculo es: {area:.2f}")

# Dibujar círculo
fig, ax = plt.subplots()

circulo = plt.Circle((0, 0), radio, fill=False)
ax.add_patch(circulo)

ax.set_xlim(-radio - 1, radio + 1)
ax.set_ylim(-radio - 1, radio + 1)
ax.set_aspect('equal')

plt.xlabel("X")
plt.ylabel("Y")
plt.title(f"Círculo de radio {radio}")
plt.grid()
plt.show()
```
