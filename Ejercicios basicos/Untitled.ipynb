```python
import matplotlib.pyplot as plt

lado = 5
area = lado ** 2

# Crear la figura
fig, ax = plt.subplots()
cuadrado = plt.Rectangle((0, 0), lado, lado, fill=True, color='skyblue', edgecolor='black')
ax.add_patch(cuadrado)

# Ajustar los límites y aspecto
ax.set_xlim(-1, lado + 1)
ax.set_ylim(-1, lado + 1)
ax.set_aspect('equal')
ax.set_title(f'Cuadrado de lado {lado} — Área: {area}')

plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

# Pedir datos al usuario
base = float(input("Ingrese la base del rectángulo: "))
altura = float(input("Ingrese la altura del rectángulo: "))

# Calcular área
area = base * altura
print("El área del rectángulo es:", area)

# Coordenadas de los vértices (para cerrar la figura se repite el primer punto al final)
x = [0, base, base, 0, 0]
y = [0, 0, altura, altura, 0]

# Graficar
plt.plot(x, y, marker='o')
plt.fill(x, y, alpha=0.3)  # opcional: rellena el rectángulo con color

# Configuración del gráfico
plt.xlabel("Base")
plt.ylabel("Altura")
plt.title(f"Rectángulo (Área = {area})")
plt.grid(True)
plt.axis("equal")

plt.show()
```
