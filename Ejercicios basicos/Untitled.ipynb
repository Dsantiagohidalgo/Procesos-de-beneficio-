```python
base = 6
altura = 4 
area = (base * altura) / 2
print(f"El área del triángulo es: {area}")
import math

a, b, c = 5, 6, 7

s = (a + b + c) / 2  # semiperímetro
area = math.sqrt(s * (s - a) * (s - b) * (s - c))
print(f"El área del triángulo es: {area:.2f}")
def area_triangulo(p1, p2, p3):
    x1, y1 = p1
    x2, y2 = p2
    x3, y3 = p3
    return abs((x1*(y2 - y3) + x2*(y3 - y1) + x3*(y1 - y2)) / 2)

area = area_triangulo((0, 0), (4, 0), (0, 3))
print(f"El área del triángulo es: {area}")

```

```python
%matplotlib inline
import matplotlib.pyplot as plt

# Datos del triángulo
base = 6
altura = 4

# Cálculo del área
area = (base * altura) / 2
print(f"El área del triángulo es: {area}")

# Vértices del triángulo (triángulo rectángulo simple con esa base y altura)
p1 = (0, 0)
p2 = (base, 0)
p3 = (0, altura)

# Dibujo
fig, ax = plt.subplots()
triangulo = plt.Polygon([p1, p2, p3], edgecolor='black', facecolor='skyblue', alpha=0.6)
ax.add_patch(triangulo)

# Etiquetas de los vértices
for p, nombre in zip([p1, p2, p3], ['A', 'B', 'C']):
    ax.plot(*p, 'ko')
    ax.text(p[0]+0.1, p[1]+0.1, nombre, fontsize=12)

# Mostrar el área en el centro del triángulo
centroide_x = (p1[0] + p2[0] + p3[0]) / 3
centroide_y = (p1[1] + p2[1] + p3[1]) / 3
ax.text(centroide_x, centroide_y, f"Área = {area}", fontsize=11, ha='center')

ax.set_xlim(-1, base + 1)
ax.set_ylim(-1, altura + 1)
ax.set_aspect('equal')
plt.grid(True)
plt.title("Triángulo (base x altura)")
plt.show()
```
