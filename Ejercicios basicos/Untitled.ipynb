```python
import matplotlib.pyplot as plt

lado = 5
area = lado ** 2

# Crear la figura
fig, ax = plt.subplots()
cuadrado = plt.Rectangle((0, 0), lado, lado, fill=True, color='skyblue', edgecolor='black')
ax.add_patch(cuadrado)

# Ajustar los límites y aspecto
ax.set_xlim(-1, lado + 1)
ax.set_ylim(-1, lado + 1)
ax.set_aspect('equal')
ax.set_title(f'Cuadrado de lado {lado} — Área: {area}')

plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

# Pedir el lado del cuadrado
lado = float(input("Ingrese el lado del cuadrado: "))

# Calcular el área
area = lado ** 2

# Mostrar el resultado
print("El área del cuadrado es:", area)

# Dibujar el cuadrado
x = [0, lado, lado, 0, 0]
y = [0, 0, lado, lado, 0]

plt.plot(x, y, marker='o')
plt.fill(x, y, alpha=0.3)

# Configuración del gráfico
plt.title("Área de un cuadrado")
plt.xlabel("Lado")
plt.ylabel("Lado")
plt.axis("equal")
plt.grid(True)

# Mostrar el gráfico
plt.show()
```
